```python
%sql
select cast(date as date) as date, item.description as description, item.quantity as quantity, item.unit_price as unit_price, item.quantity * item.unit_price as total
from (select date, explode(items) as item from main.default.json_demo) as exploded_items
where item.description in ('Monitor', 'HDMI Cable', 'Webcam')
```

```python
from pyspark.sql.functions import *

myschema = '''
    amount DOUBLE,
    id STRING,
    item_count DOUBLE,
    transaction_date TIMESTAMP,
    user_id STRING
    '''
df = spark.read.format('json').option('inferSchema', True)\
    .option('header', True)\
    .option('multiLine', False)\
    .option('timestampFormat', 'MM-dd-yyyy HH:mm:ss')\
    .schema(myschema)\
    .load('/Volumes/main/dbdemos_retail_c360/c360/orders/*')

df.withColumn('total', col('amount')*col('item_count'))\
    .withColumn('date', col('transaction_date').cast('date'))\
    .groupBy('date').agg(sum('total').alias('total_spent'))\
    .display()
```
